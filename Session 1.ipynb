#### Vectorization
* Allows for CPU-level parallelization for panel data stored in consecutive memory addresses
* Using CPU instructions to perform operations on an array of memory as a single instruction
  * BLAS, LAPACK interface - initial implementation in Fortran in 1979
  * Intel MMX instruction set, initially for music compression (.mp3)
  * Even more efficient instructions now exist, install intel mkl -> 40% boost
* Matlab first commercial success of Vectorized MMX
* Python implementation - numpy

#### Parallelization
* Multiprocessing
    * Most widely used tool for parallel processing in python
    * Creates a pool of identical processes with a copy of all of the parent process's data
    * Executes a function
    * Communication between processes using server-client via TCP socket
    * Global Interpreter Lock (GIL) - limitation
    * Not suitable for large scale jobs

#### Parallelization
* Multithreading
    * Creates pool of threads for executing a task in parallel
    * Threads share memory space - no need for copies and complicated client-server communications over TCP
    * Limitation - single interpreter, blocked by GIL
    * In practice, only useful for jobs where Python calls a long-running non-Python task to run in background
        * Classical example - IO
    * Starting in Python 3.14 GIL going away
        * brave new world...

#### Parallelization
* ray
    * Creates a distributed cluster and manages the distributed pool of workers
    * Multiprocessing interfaces available - minimal code changes necessary
    * Ease of use leads to high adoption in industry - Cubist, Millennium, Balyasny
    * Good for shardable tasks
    * Relatively resource efficient
    * Best to not use it for data - use S3 or similar.

#### Parallelization

    * as a single file
    * sharded by date
    * sharded by symbol
